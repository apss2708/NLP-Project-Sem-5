Exploratory Data Analysis on the labelled sentiment dataset used to train the sentiment classifier.

## 1. Setup & Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
import re

sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["axes.titlesize"] = 14
```

## 2. Load Dataset

```python
df = pd.read_csv("data/sample_imdb.csv")
print(df.head(10))
print(f"\nTotal reviews: {len(df)}")
print(f"Columns: {list(df.columns)}")
```

## 3. Sentiment Class Distribution

```python
class_counts = df["label"].value_counts()

fig, ax = plt.subplots()
colors = {"Positive": "#10b981", "Negative": "#ef4444", "Neutral": "#f59e0b"}
sns.barplot(x=class_counts.index, y=class_counts.values, 
            palette=[colors.get(x, "#64748b") for x in class_counts.index], ax=ax)
ax.set_title("Sentiment Class Distribution")
ax.set_xlabel("Sentiment")
ax.set_ylabel("Number of Reviews")
for i, v in enumerate(class_counts.values):
    ax.text(i, v + 0.1, str(v), ha="center", fontweight="bold")
plt.tight_layout()
plt.show()

print(class_counts)
```

## 4. Review Length Distribution

```python
df["word_count"] = df["text"].apply(lambda x: len(str(x).split()))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(df["word_count"], bins=10, color="#2563eb", ax=axes[0], edgecolor="white")
axes[0].set_title("Review Length (Word Count) Distribution")
axes[0].set_xlabel("Words per Review")
axes[0].set_ylabel("Frequency")

sns.boxplot(x=df["word_count"], color="#10b981", ax=axes[1])
axes[1].set_title("Review Length Boxplot")
axes[1].set_xlabel("Words per Review")

plt.tight_layout()
plt.show()

print(f"Min: {df['word_count'].min()} | Max: {df['word_count'].max()} | Mean: {df['word_count'].mean():.2f}")
```

## 5. Most Frequent Words by Sentiment

```python
stopwords = {"the","a","an","and","or","but","is","are","was","were","it","this","that","to","of","in","for","on","with","as","at","by","from","be","been","have","has","had","do","does","did","will","would","could","should","i","you","he","she","they","we","me","him","her","them","us","my","your","his","her","their","our","not","no","yes","so","just","very","really","quite"}

for sentiment in ["Positive", "Negative", "Neutral"]:
    texts = df[df["label"] == sentiment]["text"].astype(str).tolist()
    words = []
    for t in texts:
        words.extend(re.findall(r"\\b[a-zA-Z]{3,}\\b", t.lower()))
    words = [w for w in words if w not in stopwords]
    top = Counter(words).most_common(10)
    print(f"\n=== {sentiment} ===")
    for word, cnt in top:
        print(f"  {word}: {cnt}")
```
